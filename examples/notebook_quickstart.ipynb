```python
import pynisar
from IPython.display import Image, display
run = pynisar.process_sample('outputs', mode='dual')
pynisar.plot(run, dpi=160)
pynisar.plot_halpha(run, dpi=160)
pynisar.report(run)
display(Image(filename=str(run / 'case_study.png')))
```

## Full polarization
The quad sample contains all four measured channels. `process_sample(mode='quad')` explicitly applies the reciprocal C3 method recorded in its manifest; no extra looks are added.

```python
quad = pynisar.process_sample('outputs', mode='quad')
pynisar.plot_haalpha(quad, dpi=160)
print(quad)
```

## Your own HDF5 observation
Use `pynisar.inspect(path)` first, then `pynisar.process(path, 'outputs', channels=['HH', 'HV'], window_size=256, looks=(4, 2))`. For map products add `center=(longitude, latitude)`; for RSLC add `pixel=(row, column)`. Earthdata login is required for protected remote files.
